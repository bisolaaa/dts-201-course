```python
import statistics
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

#Number of people in the set
num_people = len(ages)
print(f"Total people: {num_people}")

#Ages
total_age = sum(ages)
print(f"Total of all ages: {total_age}")

#Mean age to 1d.p
mean_age = total_age / num_people
print(f"Mean age: {mean_age:.1f}")

#highest and lowest age
youngest = min(ages)
oldest = max(ages)
print(f"Youngest age: {youngest}, Oldest age: {oldest}")

median_age = statistics.median(ages)
print(f"Median age: {median_age}")
```

Three data points a university can collect about a student that will not help them predict their grade
1. T-Shirt Size: The student's clothing size (e.g., S, M, L, XL) collected during orientation for university merchandise. Physical body measurements have no relationship with cognitive capacity, academic preparation, or study habits. A student's physical size is a biological trait that operates completely independently of their ability to learn, retain, or apply academic information.
2. The unique hardware identification number of the student’s primary smartphone or laptop used to connect to the campus network. While network connection times might indicate when a student is online, the MAC address itself is just an arbitrary string of characters assigned by a device manufacturer. It carries no information about the student’s intellect, motivation, or the type of work they are performing on that device.
3. Chosen Campus Emergency Contact Relationship: Whether the student lists a "Mother," "Father," "Guardian," or "Spouse" as their primary emergency contact. The legal or familial relationship of an emergency contact is a logistical detail for safety protocols. It does not measure the student's personal work ethic, intelligence, or academic preparation. Two students listing a "Mother" could have completely opposite levels of academic performance.

```python
import pandas as pd
df = pd.read_excel("50_startups_messy.xlsx")

print("Shape: ", df.shape)

#to report missing counts per column
print("Missing counts: \n", df.isnull().sum())

#to report duplicate count
print("Duplicate count:", df.duplicated().sum())
```

```python
df_clean = df.drop_duplicates().copy()

#standardise state column
df_clean["State"]= df_clean["State"].str.strip().str.title()
df_clean["Marketing Spend"] = pd.to_numeric(
    df_clean["Marketing Spend"], errors = "coerce"
)
```

```python
srs_sample = df_clean.sample(n=20, random_state=7)

start_sample = df_clean.groupby("State", group_keys=False).apply(
    lambda x: x.sample(frac=0.20, random_state=7)
)
```

The simple random sample has a larger sample size, making it much closer to the true population average for both R&D and marketing spend. The simple random sample groups were also already equal.
